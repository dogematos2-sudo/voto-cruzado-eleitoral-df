```python
import zipfile
import urllib.request
import os
import pandas as pd

# Cria a pasta de trabalho se ela não existir
pasta_arquivos = "dados_candidatos"
os.makedirs(pasta_arquivos, exist_ok=True)

# 1. Baixar a base de Votação por Seção Eleitoral do DF (exemplo direto do link do TSE)
url_votacao = "https://cdn.tse.jus.br/estatistica/sead/odsele/votacao_secao/votacao_secao_2026_DF.zip"
nome_zip_votacao = os.path.join(pasta_arquivos, "votacao_df.zip")

print("Baixando os dados de votação do TSE...")
urllib.request.urlretrieve(url_votacao, nome_zip_votacao)
print("Download concluído!")

# 2. Descompactar o arquivo ZIP
with zipfile.ZipFile(nome_zip_votacao, 'r') as zip_ref:
    zip_ref.extractall(pasta_arquivos)

print("Arquivos descompactados com sucesso na pasta 'dados_candidatos'!")

# 3. Listar arquivos extraídos para confirmar
print("\nArquivos disponíveis na pasta:")
for f in os.listdir(pasta_arquivos):
    if f.endswith('.csv'):
        print(f" - {f}")
```

```python
import pandas as pd
import os

# Caminho completo do arquivo CSV descompactado
caminho_csv = os.path.join("dados_candidatos", "votacao_secao_2026_DF.csv")

print("Lendo a base de dados de votação (isso pode levar alguns segundos)...")
# Como os arquivos do TSE são grandes, usaremos um limite inicial ou a leitura otimizada
df_votos = pd.read_csv(caminho_csv, encoding='latin1', sep=';', low_memory=False)

print("\n--- Primeiras linhas da base de votação ---")
display(df_votos.head())

# Exibindo as colunas disponíveis para sabermos onde estão os votos e os nomes dos candidatos
print("\n--- Colunas disponíveis no arquivo ---")
print(df_votos.columns.tolist())
```

```python
# 1. Agrupando por Nome do Candidato/Votável e somando os votos
votos_candidato = df_votos.groupby('NM_VOTAVEL')['QT_VOTOS'].sum().reset_index()

# 2. Ordenando da maior votação para a menor
votos_candidato = votos_candidato.sort_values(by='QT_VOTOS', ascending=False)

# 3. Exibindo os 10 candidatos mais votados
print("--- Ranking de Votos por Candidato (DF) ---")
display(votos_candidato.head(10))
```

```python
# 1. Filtrar o DataFrame apenas para o cargo de Presidente (CD_CARGO geralmente é 1 para Presidente)
# Dica: Verifique se na sua coluna 'DS_CARGO' está escrito 'presidente'
df_presidente = df_votos[df_votos['DS_CARGO'].str.contains('PRESIDENTE', case=False, na=False)]

# 2. Agrupar por nome do candidato a presidente e somar os votos
votos_presidente = df_presidente.groupby('NM_VOTAVEL')['QT_VOTOS'].sum().reset_index()

# 3. Ordenar do mais votado para o menos votado
votos_presidente = votos_presidente.sort_values(by='QT_VOTOS', ascending=False)

# 4. Exibir o resultado no Colab
print("--- Votação para Presidente da República no DF ---")
display(votos_presidente)
```

```python
import zipfile
import urllib.request
import os
import pandas as pd

# Cria a pasta de trabalho se ela não existir
pasta_arquivos = "dados_candidatos"
os.makedirs(pasta_arquivos, exist_ok=True)

# 1. URL da base nacional de votação por seção para Presidente
url_votacao_br = "https://cdn.tse.jus.br/estatistica/sead/odsele/votacao_secao/votacao_secao_2026_BR.zip"
nome_zip_br = os.path.join(pasta_arquivos, "votacao_br.zip")

print("Baixando os dados nacionais de votação do TSE (isso pode demorar um pouco devido ao tamanho)...")
urllib.request.urlretrieve(url_votacao_br, nome_zip_br)
print("Download concluído!")

# 2. Descompactar o arquivo ZIP nacional
with zipfile.ZipFile(nome_zip_br, 'r') as zip_ref:
    zip_ref.extractall(pasta_arquivos)

print("Arquivos nacionais descompactados com sucesso na pasta 'dados_candidatos'!")
```

```python
import pandas as pd
import os

# Caminho do arquivo nacional descompactado
caminho_csv_br = os.path.join("dados_candidatos", "votacao_secao_2026_BR.csv")

print("Lendo a base nacional de votação...")
# Usando usecols para carregar apenas as colunas essenciais e economizar memória RAM do Colab
colunas_necessarias = ['SG_UF', 'DS_CARGO', 'NM_VOTAVEL', 'QT_VOTOS']
df_br = pd.read_csv(caminho_csv_br, encoding='latin1', sep=';', low_memory=False, usecols=colunas_necessarias)

# 1. Filtrar apenas para o Distrito Federal (DF) e para o cargo de Presidente
df_pres_df = df_br[(df_br['SG_UF'] == 'DF') & (df_br['DS_CARGO'].str.contains('PRESIDENTE', case=False, na=False))]

# 2. Agrupar e somar os votos por candidato
votos_presidente_df = df_pres_df.groupby('NM_VOTAVEL')['QT_VOTOS'].sum().reset_index()

# 3. Ordenar do mais votado para o menos votado
votos_presidente_df = votos_presidente_df.sort_values(by='QT_VOTOS', ascending=False)

# 4. Exibir o resultado no Colab
print("--- Votação para Presidente da República no Distrito Federal (DF) ---")
display(votos_presidente_df)
```

```python
import pandas as pd
import os
import plotly.express as px

# 1. Caminho do arquivo nacional descompactado
caminho_csv_br = os.path.join("dados_candidatos", "votacao_secao_2026_BR.csv")

# 2. Definindo colunas garantidas no arquivo de votação por seção
colunas_seguras = [
    'SG_UF', 'NR_ZONA', 'NR_SECAO', 'DS_CARGO',
    'NR_VOTAVEL', 'NM_VOTAVEL', 'QT_VOTOS'
]

print("Carregando base com colunas válidas...")
df_completo = pd.read_csv(caminho_csv_br, encoding='latin1', sep=';', low_memory=False, usecols=colunas_seguras)

# 3. Filtrar para o Distrito Federal e para a disputa presidencial
df_df_pres = df_completo[
    (df_completo['SG_UF'] == 'DF') &
    (df_completo['DS_CARGO'].str.contains('PRESIDENTE', case=False, na=False))
].copy()

# 4. Agrupamento Avançado por Zona Eleitoral e Candidato
df_zona = df_df_pres.groupby(['NR_ZONA', 'NM_VOTAVEL'])[['QT_VOTOS']].sum().reset_index()

print("\n--- Amostra do Agrupamento por Zona Eleitoral ---")
display(df_zona.head(10))

# 5. Dashboard Interativo: Desempenho por Zona Eleitoral (Gráfico de Barras Agrupadas)
# Filtrando os dois principais candidatos para o gráfico não ficar poluído
principais = ['FLAVIO NANTES BOLSONARO', 'LUIZ INÁCIO LULA da SILVA']
df_filtrado_grafico = df_zona[df_zona['NM_VOTAVEL'].isin(principais)]

fig_zonas = px.bar(
    df_filtrado_grafico,
    x='NR_ZONA',
    y='QT_VOTOS',
    color='NM_VOTAVEL',
    barmode='group',
    title='<b>Desempenho dos Principais Candidatos a Presidente por Zona Eleitoral (DF)</b>',
    labels={'NR_ZONA': 'Zona Eleitoral', 'QT_VOTOS': 'Total de Votos', 'NM_VOTAVEL': 'Candidato'},
    template='plotly_white'
)
fig_zonas.update_layout(xaxis=dict(type='category'))
fig_zonas.show()
```

```python
import pandas as pd
import os
import plotly.express as px

# 1. Carregar a base de Governador/Senador do DF que já estava na pasta
caminho_csv_df = os.path.join("dados_candidatos", "votacao_secao_2026_DF.csv")
cols_df = ['NR_ZONA', 'NR_SECAO', 'DS_CARGO', 'NM_VOTAVEL', 'QT_VOTOS']
df_gov_secao = pd.read_csv(caminho_csv_df, encoding='latin1', sep=';', low_memory=False, usecols=cols_df)

# Filtrar apenas para Governador
df_governador = df_gov_secao[df_gov_secao['DS_CARGO'].str.contains('GOVERNADOR', case=False, na=False)]
df_gov_zona = df_governador.groupby(['NR_ZONA', 'NM_VOTAVEL'])['QT_VOTOS'].sum().reset_index()
df_gov_zona.rename(columns={'NM_VOTAVEL': 'CANDIDATO_GOV', 'QT_VOTOS': 'VOTOS_GOV'}, inplace=True)

# 2. Preparar a base de Presidente por Zona (que já usamos)
caminho_csv_br = os.path.join("dados_candidatos", "votacao_secao_2026_BR.csv")
cols_br = ['SG_UF', 'NR_ZONA', 'NR_SECAO', 'DS_CARGO', 'NM_VOTAVEL', 'QT_VOTOS']
df_br = pd.read_csv(caminho_csv_br, encoding='latin1', sep=';', low_memory=False, usecols=cols_br)

df_pres = df_br[(df_br['SG_UF'] == 'DF') & (df_br['DS_CARGO'].str.contains('PRESIDENTE', case=False, na=False))]
df_pres_zona = df_pres.groupby(['NR_ZONA', 'NM_VOTAVEL'])['QT_VOTOS'].sum().reset_index()
df_pres_zona.rename(columns={'NM_VOTAVEL': 'CANDIDATO_PRES', 'QT_VOTOS': 'VOTOS_PRES'}, inplace=True)

# 3. Cruzar (Merge) os dados por Zona Eleitoral
df_cruzado = pd.merge(df_pres_zona, df_gov_zona, on='NR_ZONA')

print("--- Amostra do Cruzamento de Votos por Zona Eleitoral ---")
display(df_cruzado.head(10))

# 4. Criando um Gráfico de Dispersão (Scatter Plot) interativo para ver a correlação
fig_cruzado = px.scatter(
    df_cruzado,
    x='VOTOS_PRES',
    y='VOTOS_GOV',
    color='CANDIDATO_PRES',
    symbol='CANDIDATO_GOV',
    hover_data=['NR_ZONA'],
    title='<b>Correlação de Votos: Presidente vs Governador por Zona Eleitoral (DF)</b>',
    labels={'VOTOS_PRES': 'Votos para Presidente', 'VOTOS_GOV': 'Votos para Governador'},
    template='plotly_white'
)
fig_cruzado.show()
```

```python
import pandas as pd
import os
import plotly.express as px

# 1. Carregar a base de Governador do DF
caminho_csv_df = os.path.join("dados_candidatos", "votacao_secao_2026_DF.csv")
cols_df = ['NR_ZONA', 'NR_SECAO', 'DS_CARGO', 'NM_VOTAVEL', 'QT_VOTOS']
df_gov_secao = pd.read_csv(caminho_csv_df, encoding='latin1', sep=';', low_memory=False, usecols=cols_df)

# Filtrar Governador e somar por Zona
df_governador = df_gov_secao[df_gov_secao['DS_CARGO'].str.contains('GOVERNADOR', case=False, na=False)]
df_gov_zona = df_governador.groupby(['NR_ZONA', 'NM_VOTAVEL'])['QT_VOTOS'].sum().reset_index()
df_gov_zona.rename(columns={'NM_VOTAVEL': 'CANDIDATO_GOV', 'QT_VOTOS': 'VOTOS_GOV'}, inplace=True)

# 2. Carregar a base de Presidente (DF)
caminho_csv_br = os.path.join("dados_candidatos", "votacao_secao_2026_BR.csv")
cols_br = ['SG_UF', 'NR_ZONA', 'NR_SECAO', 'DS_CARGO', 'NM_VOTAVEL', 'QT_VOTOS']
df_br = pd.read_csv(caminho_csv_br, encoding='latin1', sep=';', low_memory=False, usecols=cols_br)

df_pres = df_br[(df_br['SG_UF'] == 'DF') & (df_br['DS_CARGO'].str.contains('PRESIDENTE', case=False, na=False))]
df_pres_zona = df_pres.groupby(['NR_ZONA', 'NM_VOTAVEL'])['QT_VOTOS'].sum().reset_index()
df_pres_zona.rename(columns={'NM_VOTAVEL': 'CANDIDATO_PRES', 'QT_VOTOS': 'VOTOS_PRES'}, inplace=True)

# 3. Cruzar (Merge) os dados por Zona Eleitoral
df_cruzado = pd.merge(df_pres_zona, df_gov_zona, on='NR_ZONA')

# 4. Filtrar apenas os principais candidatos para focar a análise e o gráfico
principais_pres = ['FLAVIO NANTES BOLSONARO', 'LUIZ INÁCIO LULA DA SILVA']
principais_gov = ['CELINA LEAO HIZIM FERREIRA', 'LEANDRO ANTÔNIO GRASS PEIXOTO']

df_foco = df_cruzado[
    df_cruzado['CANDIDATO_PRES'].isin(principais_pres) &
    df_cruzado['CANDIDATO_GOV'].isin(principais_gov)
].copy()

print("--- Amostra do Cruzamento Focado (Governador vs Presidente) ---")
display(df_foco.head(10))

# 5. Criando um Gráfico de Dispersão (Scatter Plot) limpo e profissional
fig_foco = px.scatter(
    df_foco,
    x='VOTOS_PRES',
    y='VOTOS_GOV',
    color='CANDIDATO_PRES',
    symbol='CANDIDATO_GOV',
    hover_data=['NR_ZONA'],
    title='<b>Cruzamento de Votos: Principais Nomes para Presidente vs Governador (DF)</b>',
    labels={
        'VOTOS_PRES': 'Votos para Presidente',
        'VOTOS_GOV': 'Votos para Governador',
        'CANDIDATO_PRES': 'Candidato a Presidente',
        'CANDIDATO_GOV': 'Candidato a Governador'
    },
    template='plotly_white'
)
fig_foco.show()
```
